```python
# W# Read Bronze Delta Tables

df_customers = spark.read.table("bronze_customers")
df_orders = spark.read.table("bronze_orders")
df_order_items = spark.read.table("bronze_order_items")
df_products = spark.read.table("bronze_products")
df_returns = spark.read.table("bronze_returns")
df_sales_targets = spark.read.table("bronze_sales_targets")

```

```python
display(df_customers)
display(df_orders)
display(df_order_items)
display(df_products)
display(df_returns)
display(df_sales_targets)
```

```python
# Remove duplicate customers

df_customers = df_customers.dropDuplicates(["CustomerID"])
```

```python
from pyspark.sql.functions import to_date

df_orders = df_orders.withColumn(
    "OrderDate",
    to_date("OrderDate")
)
```

```python
df_products = df_products.na.drop(subset=["ProductID"])
```

```python
df_returns = df_returns.fillna("Unknown")
```

```python
display(df_customers)
display(df_orders)
display(df_products)
display(df_returns)
```

```python
df_customers.write.mode("overwrite").format("delta").saveAsTable("silver_customers")

df_orders.write.mode("overwrite").format("delta").saveAsTable("silver_orders")

df_order_items.write.mode("overwrite").format("delta").saveAsTable("silver_order_items")

df_products.write.mode("overwrite").format("delta").saveAsTable("silver_products")

df_returns.write.mode("overwrite").format("delta").saveAsTable("silver_returns")

df_sales_targets.write.mode("overwrite").format("delta").saveAsTable("silver_sales_targets")
```
